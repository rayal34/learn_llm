```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
with open("names.txt", "r") as f:
    words = f.read().splitlines()
```

```python
chars = sorted(list(set("".join(words))))
stoi, itos = {}, {}
for i, s in enumerate(["."] + sorted(list(chars))):
    stoi[s] = i
    itos[i] = s
```

```python
# create dataset
xs, ys = [], []

for word in words:
    chs = ["."] + list(word) + ["."]
    for ch1, ch2 in zip(chs, chs[1:]):
        idx1 = stoi[ch1]
        idx2 = stoi[ch2]
        xs.append(idx1)
        ys.append(idx2)

xs = torch.tensor(xs)
ys = torch.tensor(ys)

num = xs.nelement()
num_classes = len(stoi)
W = torch.randn((27, num_classes), requires_grad=True)
```

```python
# gradient descent
for k in range(100):
    xenc = F.one_hot(xs, num_classes=num_classes).float()

    # forward pass
    logits = xenc @ W
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(num), ys].log().mean() + 0.01 * (W ** 2).mean()
    print(loss.item())

    # backward pass
    W.grad = None
    loss.backward()

    # update step
    W.data += -50 * W.grad
```

```python
def sample_from_model(W, stoi, itos):
    word = []
    char = "."
    special_char = "."

    while True:
        xs = torch.tensor([stoi[char]])
        xenc = F.one_hot(xs, num_classes=num_classes).float()

        logits = xenc @ W
        counts = logits.exp()
        probs = counts / counts.sum(1, keepdim=True)

        idx = torch.multinomial(probs, 1, replacement=True)
        char = itos[idx.item()]
        if char == special_char:
            break
        word.append(char)


    return "".join(word)
```

```python
for _ in range(10):
    print(sample_from_model(W, stoi, itos))
```
